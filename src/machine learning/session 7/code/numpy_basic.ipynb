```python
def convert_dtype(arr, dtype):
    try:
        arr = np.array(arr, dtype=dtype)
        return arr.astype(dtype)
    except ValueError as e:
        raise ValueError(f"invalid target data type: {dtype}. Details: {e}")
    except TypeError as e:
            raise TypeError(f"invalid target data type: {dtype}. Details: {e}")
    

    
```

```python
lst = [1, 2, 3, 4]
convert_dtype(lst, float)
 
```

```python
def check_dtype(arr, dtype):
    
    if not isinstance(arr, np.ndarray):
        raise TypeError("Input must be a numpy array.")
        
    dtype = np.dtype(dtype)
    return arr.dtype == dtype   
 
```

```python
lst = [1, 2, 3, 4]
convert_dtype(lst, float)
```

```python
def init_weights(n_feature, method = "zeros", seed =None): # lw sora msln kol pixel have weght 
    if not isinstance(n_feature, int):
        raise TypeError("n_feature must be an integer.")
    if n_feature <= 0:
        raise ValueError("n_feature must be a positive integer.")
    if method=="zeros":
        return np.zeros(n_feature)
    if method=="randoms":
        if seed is not None:
            np.random.seed(seed)
        return np.random.rand(n_feature)*0.01
    raise ValueError(f"un supported method: {method}. Supported methods are 'zeros' and 'randoms'.")
      
```

```python
init_weights(5,"randoms")
```
