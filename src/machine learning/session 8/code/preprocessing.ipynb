# 6. Check Nulls
# 7. Handle Nulls

```python
df.isnull()
```

```python
null = df.isnull().sum()
print(null)
ratio = (null / 891) * 100
ratio
```

```python
# for example you're in a company and want to decide from the three cols: age, cabine, and embarked
# most of values exist (about 30%) use the mode in case of a categorical data
# otherwise, if most of the data is missing, drop the column
# embarked has only 2 rows that has missing values, you can use the mode, or drop the two rows
# you can even link a column by another column and  produce a prdiction according to the other linked columns
```

```python
df = df.dropna(subset=['Embarked']) #dropna to drop the rows that have null values in the Embarked column. This is because the number of rows with null values in the Embarked column is very small (only 2 rows), so dropping them will not affect the dataset significantly.
```

```python
df = df.drop("Cabin", axis = 1)
```

```python
df.describe() # to get the statistical summary of the numerical columns in the df. It helps us to understand the data better and to decide which columns to drop or keep.
```

```python
df.describe()
median = df['Age'].median()
print(median)
```

```python
df['Age'] = df['Age'].fillna(median, inplace= True) # fill the null values in the Age column with the median value. This is a common approach when dealing with missing data in numerical columns.
```

```python
null = df.isnull().sum()
ratio = (null / 891) * 100
pd.DataFrame({"null": null, "ratio": ratio})
```

# 8. Check Outliers
# 9. Handle Outliers

IQR for handling outliers
it draws the safe zone of the data 


```python
num_col = df.select_dtypes("number").columns
num_col
```

```python
num_col = df.select_dtypes("number").columns
num_col
for col in num_col:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    print(Q1, Q3)
    IQR = Q3 - Q1
    print("IQR:", IQR)
    lower_fence = Q1 - 1.5 * IQR
    upper_fence = Q3 + 1.5 * IQR
    print("Lower fence:", lower_fence, "Upper fence:", upper_fence)
```
