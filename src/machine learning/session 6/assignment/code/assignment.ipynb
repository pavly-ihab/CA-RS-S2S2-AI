```python
class BankAccount:
    """
    A class to represent and manage a bank account.

    Attributes:
        balance (float): The current money available in the account.
    """

    def __init__(self, initial_balance: float = 0.0) -> None:
        """
        Initializes the BankAccount with a starting balance.

        Args:
            initial_balance (float): The amount to start the account with. 
                                     Defaults to 0.0. If negative, defaults to 0.0.
        """
        # Validate initial balance
        if initial_balance < 0:
            print("Initial balance cannot be negative. Setting balance to 0.")
            self.balance: float = 0.0
        else:
            self.balance: float = float(initial_balance)

    def deposit(self, amount: float) -> None:
        """
        Adds money to the account balance.

        Args:
            amount (float): The amount of money to deposit. Must be greater than zero.
        """
        # Validate deposit amount
        if amount <= 0:
            print("Deposit amount must be greater than zero.")
            return
        
        self.balance += amount
        print(f'Deposited {amount}. New balance: {self.balance}')

    def withdraw(self, amount: float) -> None:
        """
        Deducts money from the account balance.

        Args:
            amount (float): The amount of money to withdraw. Must be greater than zero
                            and less than or equal to the current balance.
        """
        # Validate withdrawal amount
        if amount <= 0:
            print("Withdrawal amount must be greater than zero.")
            return
        
        if amount > self.balance:
            print('Insufficient funds.')
        else:
            self.balance -= amount
            print(f'Withdrew {amount}. New balance: {self.balance}')

    def check_balance(self) -> None:
        """
        Prints the current available balance to the console.
        """
        print(f'Current balance: {self.balance}')

# Usage demonstration
if __name__ == "__main__":
    # 1. Initialize the account
    account = BankAccount(100.0)
    
    # 2. Perform valid operations
    account.deposit(50.0)
    account.withdraw(30.0)
    account.check_balance()
    
    print("-" * 30)
    
    # 3. Demonstrate the new input validation
    account.deposit(-20.0)      # Invalid deposit
    account.withdraw(-10.0)     # Invalid withdrawal
    account.withdraw(500.0)     # Insufficient funds
```

```python
class Calculator:
    def __init__(self):
        """Initializes the calculator class."""
        pass  # No initialization logic needed for a simple calculator

    def add(self, a: float, b: float) -> float:
        """Add two numbers."""
        return a + b

    def subtract(self, a: float, b: float) -> float:
        """Subtract two numbers."""
        return a - b

    def multiply(self, a: float, b: float) -> float:
        """Multiply two numbers."""
        return a * b

    def divide(self, a: float, b: float):
        """Divide two numbers. Returns a string error if dividing by zero."""
        if b == 0:
            return "Error: Cannot divide by zero."
        return a / b

    def calculator(self):
        """Main calculator logic to get user input and perform operations."""
        print("Welcome to the calculator!")
        print("Choose an operation:")
        print("1: Add")
        print("2: Subtract")
        print("3: Multiply")
        print("4: Divide")

        choice = input("Enter the number of the operation you want to perform (1/2/3/4): ")

        if choice in ['1', '2', '3', '4']:
            try:
                num1 = float(input("Enter the first number: "))
                num2 = float(input("Enter the second number: "))
            except ValueError:
                print("Invalid input! Please enter numbers only.")
                return
            
            if choice == '1':
                print(f"{num1} + {num2} = {self.add(num1, num2)}")
            elif choice == '2':
                print(f"{num1} - {num2} = {self.subtract(num1, num2)}")
            elif choice == '3':
                print(f"{num1} * {num2} = {self.multiply(num1, num2)}")
            elif choice == '4':
                result = self.divide(num1, num2)
                if isinstance(result, str):
                    print(result)  # Prints error message
                else:
                    print(f"{num1} / {num2} = {result}")
        else:
            print("Invalid choice! Please select a valid operation (1/2/3/4).")

# Usage example
if __name__ == "__main__":
    calc = Calculator()
    calc.calculator()
```

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass
    
    def describe(self):
        print(f"I am a {type(self).__name__}. I am a living creature.")
        
class Dog(Animal):
        def make_sound(self):
            return "Woof!"
      
class Cat(Animal):
        def make_sound(self):
            return "Meow!"
             
class Cow(Animal):
        def make_sound(self):
            return "Moo!"
        
# Usage example
if __name__ == "__main__":
    dog = Dog()
    cat = Cat()
    cow = Cow()
    
    
    
    for animal in [dog, cat, cow]:
        animal.describe()
        print(f"Sound: {animal.make_sound()}")
        print("-" * 30)           
```

```python
class TextFileReader:
    def __init__(self, file_path: str):
        """
        Initializes the TextFileReader with the path to the text file.

        Args:
            file_path (str): The path to the text file to be read.
        """
        self.file_path = file_path
        self.content = None
        
    def read_file(self):
            # 'r' stands for read mode, and we use 'utf-8' encoding to handle a wide range of characters.
            with open(self.file_path, 'r', encoding='utf-8') as file:
                self.content = file.read()
                
    def count_lines(self) -> int:
            """
            Counts the number of lines in the text file.

            Returns:
                int: The number of lines in the file.
            """
            if self.content is not None:
                # splitlines() safely handles different newline characters (\n, \r\n)
                return len(self.content.splitlines())
            return 0  # If content is None, return 0 lines
        
    def count_words(self) -> int:
            """
            Counts the number of words in the text file.

            Returns:
                int: The number of words in the file.
            """
            if self.content is not None:
                # split() splits on any whitespace and handles multiple spaces
                return len(self.content.split())
            return 0  # If content is None, return 0 words
        
    def count_characters(self) -> int:
            """
            Counts the number of characters in the text file.

            Returns:
                int: The number of characters in the file.
            """
            if self.content is not None:
                return len(self.content)
            return 0  # If content is None, return 0 characters
        
    def display_content(self) -> None:
            """
            Displays the content of the text file to the console.
            """
            if self.content is not None:
                print(self.content)
            else:
                print("No content to display. Please read the file first.")
                        
                        
if __name__ == "__main__":
    # 1. Create a temporary sample file to test with
    test_file = "sample.txt"
    sample_text = "Hello there!\nThis is a sample text file.\nIt has three lines."
    with open(test_file, "w", encoding="utf-8") as f:
        f.write(sample_text)

    # 2. Instantiate the class with the file path
    reader = TextFileReader(test_file)

    # 3. Read the file and execute all methods
    reader.read_file()
    reader.display_content()
    
    print(f"Lines: {reader.count_lines()}")
    print(f"Words: {reader.count_words()}")
    print(f"Characters: {reader.count_characters()}")
    
```

```python
def get_word_frequencies(words_list):
    unique_words = set(words_list)
    counts = {word: 0 for word in unique_words}
    
    for i in unique_words:
        for j in words_list:
            if i == j:
                counts[i] += 1
                
    return counts

# Test the function
sample_words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
result = get_word_frequencies(sample_words)
print(result)
```

```python
def read_txt_file(file_path):
    """Reads the contents of a text file and returns it as a string."""
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()
        return content
    except FileNotFoundError:
        return f"Error: The file '{file_path}' was not found."
    except IOError:
        return "Error: An error occurred while reading the file."

# Usage example
if __name__ == "__main__":
    file_path = r"C:\Users\bavly\OneDrive\Desktop\depi_cairo_AI\CA-RS-S2S2-AI\src\machine learning\session 6\assignment\code\sample.txt"  # Replace with your file path
    file_content = read_txt_file(file_path)
    print(file_content)
```
