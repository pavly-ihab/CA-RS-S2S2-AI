```python
class Multiplication_table:
    """
    A class to generate and print multiplication tables.

    Attributes:
        number: The number for which to generate the multiplication table.
        limit: The range of the multiplication table (default is 10).
    """

    def __init__(self, number: int, limit: int = 10) -> None:
        """
        Initialize the MultiplicationTable with a number and a limit.

        Args:
            number: The number for which to generate the multiplication table.
            limit: The range of the multiplication table (default is 10).
        """
        self.number = number
        self.limit = limit

    def print_table(self) -> None:
        """
        Print the multiplication table for the given number.

        Returns:
            None

        Example:
            >>> table = MultiplicationTable(5)
            >>> table.print_table()
            5 x 1 = 5
            5 x 2 = 10
            5 x 3 = 15
            ...
            5 x 10 = 50
        """
        for i in range(1, self.limit + 1):
            print(f"{self.number} x {i} = {self.number * i}")


# Example usage
if __name__ == "__main__":
    table = Multiplication_table(11, 10)  # multiplication table for 9 up to 15
    table.print_table()
```

```python
class Pair_prime:
    """
    Task1 class handles finding twin prime pairs below a given limit.

    Twin primes are pairs of consecutive odd numbers that are both prime,
    e.g. (3, 5), (5, 7), (11, 13).

    Attributes:
        limit (int): The upper bound (exclusive) for the search range.
    """

    def __init__(self, limit: int) -> None:
        """
        Initialize Task1 with the search limit.

        Args:
            limit (int): The upper bound (exclusive) for the search range.
        """
        self.limit = limit

    def is_prime(self, n: int) -> bool:
        """
        Check whether a number is prime.

        Args:
            n (int): The number to check.

        Returns:
            bool: True if n is prime, False otherwise.
        """
        if n < 2:
            return False
        for i in range(2, n):
            if n % i == 0:
                return False
        return True

    def find_twin_primes(self) -> list:
        """
        Find all twin prime pairs below the limit.

        Returns:
            list: A list of tuples, each containing a pair of twin primes.
        """
        twin_primes = []
        for num in range(2, self.limit - 2):
            if self.is_prime(num) and self.is_prime(num + 2):
                twin_primes.append((num, num + 2))
        return twin_primes


# Example usage
result = Pair_prime(1000)
print(result.find_twin_primes())
```

```python
class Prime_factors:
    """
    Task2 class handles finding the prime factors of a number.

    Attributes:
        number (int): The number to factorize.
    """

    def __init__(self, number: int) -> None:
        """
        Initialize Task2 with the number to factorize.

        Args:
            number (int): The number to factorize.
        """
        self.number = number

    def get_prime_factors(self) -> list:
        """
        Find the prime factors of the number.

        Returns:
            list: A list of prime factors, in ascending order (with repetition).

        Example:
            Task2(56).get_prime_factors() -> [2, 2, 2, 7]
        """
        n = self.number
        factors = []
        for i in range(2, n + 1):
            while n % i == 0:
                factors.append(i)
                n //= i
        return factors


# Example usage
factors = Prime_factors(56)
print(factors.get_prime_factors())
```

```python
class decimal_to_binary:
    """
    Task3 class handles converting a decimal number to its binary representation.

    Attributes:
        number (int): The decimal number to convert.
    """

    def __init__(self, number: int) -> None:
        """
        Initialize Task3 with the decimal number to convert.

        Args:
            number (int): The decimal number to convert.
        """
        self.number = number

    def to_binary(self) -> str:
        """
        Convert the decimal number to its binary representation.

        Returns:
            str: A string representing the binary form of the number.

        Example:
            Task3(10).to_binary() -> '1010'
        """
        return bin(self.number)[2:]


# Example usage
result = decimal_to_binary(10)
print(result.to_binary())
```

```python
class perfect_numbers:
    """
    Task4 class handles finding all perfect numbers within a given range.

    A perfect number is a number whose proper divisors (excluding itself)
    sum up to the number itself, e.g. 28 = 1 + 2 + 4 + 7 + 14.

    Attributes:
        start (int): The lower bound of the range (inclusive).
        end (int): The upper bound of the range (inclusive).
    """

    def __init__(self, start: int, end: int) -> None:
        """
        Initialize Task4 with the range to search.

        Args:
            start (int): The lower bound of the range (inclusive).
            end (int): The upper bound of the range (inclusive).
        """
        self.start = start
        self.end = end

    def find_perfect_numbers(self) -> list:
        """
        Find all perfect numbers within the range.

        Returns:
            list: A list of perfect numbers found within the range.
        """
        perfect_numbers = []
        for num in range(self.start, self.end + 1):
            divisor_sum = 0
            for i in range(1, num):
                if num % i == 0:
                    divisor_sum += i
            if divisor_sum == num and num > 0:
                perfect_numbers.append(num)
        return perfect_numbers


# Example usage
result = perfect_numbers(1, 10000)
print(result.find_perfect_numbers())
```
