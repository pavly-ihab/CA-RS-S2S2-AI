# <a id="3"> 3 ) What is the distribution for users' ages? </a>

```python
plt.figure(figsize=(8,5))

bin_edges = np.arange(df['member_age'].min(), df['member_age'].max()+2, 2)
plt.hist(data=df, x='member_age', bins=bin_edges)

plt.xlabel('Age (years)')
plt.ylabel('Number of Trips')
plt.title("Distribution of Users' Ages")


plt.show()

df['member_age'].describe()
```

# <a id="4">  4 ) . Do the users' ages have outlier values?</a>


```python
# Check for outliers using a boxplot, then zoom the histogram on the typical range
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='member_age', color=sns.color_palette()[2])
plt.xlabel('Age (years)')
plt.title("Boxplot of Users' Ages (checking for outliers)")
plt.show()

plt.figure(figsize=(8,5))
bin_edges = np.arange(15, 80+2, 2)
plt.hist(data=df, x='member_age', bins=bin_edges)
plt.xlim(15, 80)
plt.xlabel('Age (years)')
plt.ylabel('Number of Trips')
plt.title("Distribution of Users' Ages (zoomed to 15-80)")
plt.show()

print('Number of users older than 80: {}'.format((df['member_age'] > 80).sum()))
print('Oldest recorded age: {}'.format(df['member_age'].max()))
```

- **370 users** are recorded as older than 80, which is a small but noticeable group of outliers pulling the distribution's tail to the right.

# <a id="8"> 8) . What the average duration of trip for the categories of user type?</a>

```python
# Average trip duration (minutes) by user type
duration_by_usertype = df.groupby('user_type')['duration_min'].mean()
print(duration_by_usertype)

plt.figure(figsize=(8,5))
sns.barplot(x=duration_by_usertype.index, y=duration_by_usertype.values, color=sns.color_palette()[0])
plt.xlabel('User Type')
plt.ylabel('Average Trip Duration (minutes)')
plt.title('Average Trip Duration by User Type')
plt.show()
```

# Make visualization to easy get insight


> ### <mark>Your analsys and exploration here </mark>

- **Customers** take much longer trips on average (about **23.9 minutes**) than **Subscribers** (about **10.7 minutes**).
- This makes sense: Subscribers likely use the bikes for quick, routine commuting trips, while Customers (casual/one-time users) may be using bikes more for leisure or exploring, resulting in longer rides.

# <a id="1100"> Conclusions </a>

- **Trip duration is the clearest differentiator by user type**: Customers ride roughly twice as long on average (~24 min) as Subscribers (~11 min), suggesting Subscribers use the bikes for quick commuting while Customers use them more for leisure/exploration.

- Most trips are **short**, typically under 15 minutes, and the duration distribution is strongly right-skewed with a long tail of much longer rides.

## limitation

- We got missing data (mainly in `member_birth_year`, `member_gender`, and station id/name columns), which we handled by imputing the median (numeric) or mode/'Other' (categorical) — this is a simplification that can introduce some bias, especially visible in the age outliers (up to 141 years).

- The `start_time`/`end_time` columns in this extract did not contain usable full timestamps, so time-of-day/day-of-week patterns could not be explored in this analysis.
