```python
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#Data
x = np.array([[1],[2],[3],[4]])
y = np.array([2,4,6,8])
#linear Regression
lin_reg = LinearRegression()
lin_reg.fit(x,y)
y_pred_lin = lin_reg.predict(x)# wx+b
#Ridge Regression
ridge = Ridge(alpha= 10.0)
ridge.fit(x,y)
y_pred_ridge = ridge.predict(x)
#lasso Regression
lasso = Lasso(alpha= 10.0)
lasso.fit(x,y)
y_pred_lasso = lasso.predict(x)

print("Linear Regression:")
print("MSE: ", mean_squared_error(y,y_pred_lin))
print("R2 Score: ", r2_score(y,y_pred_lin))
print("#"*50)

print("Ridge Regression:")
print("MSE: ", mean_squared_error(y,y_pred_ridge))
print("R2 Score: ", r2_score(y,y_pred_ridge))
print("#"*50)

print("Lasso Regression:")
print("MSE: ", mean_squared_error(y,y_pred_lasso))
print("R2 Score: ", r2_score(y,y_pred_lasso)) # help for feature selection, it can shrink some coefficients to zero, effectively removing them from the model.
print("#"*50)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,5))
plt.subplot(1,3,1)
plt.scatter(x,y, color= 'blue', label= 'Data')
plt.plot(x,y_pred_lin, color= 'red', label= 'Linear Regression')
plt.title('Linear Regression')
plt.legend()

plt.subplot(1,3,2)
plt.scatter(x,y, color= 'blue', label= 'Data')
plt.plot(x,y_pred_ridge, color= 'green', label= 'Ridge Regression')
plt.title('Ridge Regression')
plt.legend()

plt.subplot(1,3,3)
plt.scatter(x,y, color= 'blue', label= 'Data')
plt.plot(x,y_pred_lasso, color= 'orange', label= 'Lasso Regression')
plt.title('Lasso Regression')
plt.legend()
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,3,1)
plt.scatter(x,y, color= 'blue', label= 'Data')
plt.plot(x,y_pred_lin, color= 'red', label= 'Linear Regression')
plt.plot(x,y_pred_ridge, color= 'green',linestyle='--', label= 'Ridge Regression')
plt.plot(x,y_pred_lasso, color= 'orange',linestyle='-.', label= 'Lasso Regression')
plt.title('Linear Regression, Ridge Regression, Lasso Regression')
plt.legend()
```
