```python
def print_multiplication_table(number: int, limit: int = 10) -> None: # the function takes two parameters: number (the number for which to generate the multiplication table) and limit (the range of the multiplication table, default is 10). It prints the multiplication table for the given number up to the specified limit.
    """
    Print the multiplication table for a given number.
    
    Args:
        number: The number for which to generate the multiplication table.
        limit: The range of the multiplication table (default is 10).
    
    Returns:
        None
    
    Example:
        >>> print_multiplication_table(5)
        5 x 1 = 5
        5 x 2 = 10
        5 x 3 = 15
        ...
        5 x 10 = 50
    """
    for i in range(1, limit + 1):
        print(f"{number} x {i} = {number * i}")


# Example usage
if __name__ == "__main__":
    print_multiplication_table(9,15)  # This will print the multiplication table for 9 up to 15
```

```python
from typing import List, Tuple


def find_twin_primes(limit: int) -> List[Tuple[int, int]]:
    """
    Print and return twin prime pairs less than the given limit.

    Twin primes are pairs of consecutive odd numbers that are both prime,
    e.g. (3, 5), (5, 7), (11, 13).

    Args:
        limit: The upper bound (exclusive) for the search.

    Returns:
        A list of twin prime pairs found below the limit.
    """

    def is_prime(n: int) -> bool:
        """Check if a number is prime."""
        if n < 2:
            return False
        for i in range(2, n):
            if n % i == 0:
                return False
        return True

    twin_primes: List[Tuple[int, int]] = []

    for num in range(2, limit - 2):
        if is_prime(num) and is_prime(num + 2):
            twin_primes.append((num, num + 2))

    return twin_primes


# Example usage
pairs = find_twin_primes(1000)
for p in pairs:
    print(p)

print("Total pairs:", len(pairs))
```

```python
def prime_factors(n: int) -> list:
    """
    Find the prime factors of a number.

    Example:
        prime_factors(56) -> [2, 2, 2, 7]

    Args:
        n: The number to factorize.

    Returns:
        A list of prime factors, in ascending order (with repetition).
    """
    factors = []

    for i in range(2, n + 1):
        while n % i == 0:
            factors.append(i)
            n //= i

    return factors


# Example usage
print(prime_factors(56))   # [2, 2, 2, 7]
print(prime_factors(100))  # [2, 2, 5, 5]
print(prime_factors(97))   # [97]
```

```python
def decimal_to_binary(n: int) -> str:
    """
    Convert a decimal number to its binary representation.

    Example:
        decimal_to_binary(10) -> '1010'

    Args:
        n: The decimal number to convert.

    Returns:
        A string representing the binary form of the number.
    """
    return bin(n)[2:]


# Example usage
print(decimal_to_binary(10))   # '1010'
print(decimal_to_binary(56))   # '111000'
print(decimal_to_binary(0))    # '0'
```

```python
def find_perfect_numbers(start: int, end: int) -> None:
    """
    Print all perfect numbers within a given range.

    A perfect number is a number whose proper divisors (excluding itself)
    sum up to the number itself.

    Example:
        28 is perfect since 1 + 2 + 4 + 7 + 14 = 28.

    Args:
        start: The lower bound of the range (inclusive).
        end: The upper bound of the range (inclusive).
    """
    for num in range(start, end + 1):
        sum_of_divisors = 0

        for i in range(1, num):
            if num % i == 0:
                sum_of_divisors += i

        if sum_of_divisors == num:
            print(num)


# Example usage
find_perfect_numbers(1, 10000)
```
