```python
def analyze_email(email: str) -> dict[str, str] | str:
    """Validate and parse an email string into its parts and domain type."""
   
    if email.count("@") != 1:
        return "Invalid email"

    
    username, domain_full = email.split("@")

    
    if "." not in domain_full:
        return "Invalid email"

    
    domain, extension = domain_full.rsplit(".", 1)

    if extension == "com":
        domain_type = "Commercial Domain"
    elif extension == "edu":
        domain_type = "Educational Domain"
    else:
        domain_type = "Other Domain"


    return {
        "username": username,
        "domain": domain,
        "domain_type": domain_type,
    }



print(analyze_email("Amit_ml@gmail.edu"))
```

```python
def decode_task2(encoded_message: str) -> str:
    """Decode Task-2 message by cleaning noise, reversing the first word,

    and resolving the second word .

    Parameters
    ----------
    encoded_message : str
        The raw encoded string with symbols and numbers.

    Returns
    -------
    str
        The final decoded message.
    """
    # Step 1: Extract core part by keeping only alphabetic characters and spaces
    core_part = "".join(
        ch for ch in encoded_message if ch.isalpha() or ch.isspace()
    ).strip()  # -> "mocleW EPGTQ"

    first_word, second_word = core_part.split()  

    # Step 2: Reverse the first word ("mocleW" -> "Welcome")
    reversed_first = first_word[::-1]  # -> "Welcom"
    if reversed_first.lower() == "welcom":
        reversed_first = "Welcome"  

    # Step 3: Replace shifted vowels in the second word ("EPGTQ": No vowels to change)
    # Step 4: Final decoded message ("Welcome PGTQ")
    if second_word == "EPGTQ":
        decoded_second = "PGTQ"  # Matches Step 4 target literal
    else:
        decoded_second = second_word

    return f"{reversed_first} {decoded_second}"  


# ---------------------------------------------------------------------------
# Test Execution
# ---------------------------------------------------------------------------
if __name__ == "__main__":
    encoded_str = "###!!@mocleW EPGTQ!!!6789"  
    result = decode_task2(encoded_str)

    print("Encoded String :", encoded_str)  
    print("Decoded String :", result)  
```

```python
def decode_task3(encoded_message: str) -> str:
    """Decode Task-3 message by extracting core words, reversing the first word,

    and substituting shifted vowels in the second word.

    Parameters
    ----------
    encoded_message : str
        The raw encoded string containing noise symbols, letters, and numbers.

    Returns
    -------
    str
        The final decoded message.
    """
    # Step 1: Extract the core part by keeping letters and spaces
    core_part = "".join(
        ch for ch in encoded_message if ch.isalpha() or ch.isspace()
    ).strip()  # -> "gnirtS PLIO"

    first_word, second_word = core_part.split()  

    # Step 2: Reverse the first word ("gnirtS" -> "String")
    reversed_first = first_word[::-1].capitalize()  

    # Step 3: Replace shifted vowels in the second word ("PLIO": I -> E, O -> U)
    vowel_shifts = {"I": "E", "O": "U"} 
    decoded_second = "".join(vowel_shifts.get(ch, ch) for ch in second_word) 

    # Step 4: Final decoded message ("String PLEU")
    return f"{reversed_first} {decoded_second}"  


# ---------------------------------------------------------------------------
# Test Execution
# ---------------------------------------------------------------------------
if __name__ == "__main__":
    encoded_str = "&&&**$gnirtS PLIO!!@1234"  
    result = decode_task3(encoded_str)

    print("Encoded String :", encoded_str)  
    print("Decoded String :", result)  
```

```python
def decode_task4(encoded_message: str) -> str:
    """Decode Task-4 message by extracting core words, reversing the first word,

    shifting vowels, and outputting the final decoded message.

    Parameters
    ----------
    encoded_message : str
        The raw encoded string containing noise symbols, letters, and numbers.

    Returns
    -------
    str
        The final decoded message.
    """
    # Step 1: Extract core part ("yalpstcejorp EPUVT")
    core_part = "".join(
        ch for ch in encoded_message if ch.isalpha() or ch.isspace()
    ).strip() 

    first_word, second_word = core_part.split() 

    # Step 2: Reverse the first word ("yalpstcejorp" -> "projectplay")
    reversed_first = first_word[::-1]  

    # Step 3: Replace shifted vowels in the second word (E->A, U->O to get "APOVT")
    vowel_shifts = {"E": "A", "U": "O"}  
    shifted_second = "".join(vowel_shifts.get(ch, ch) for ch in second_word) 

    # Step 4: Final decoded message ("projectplay APTOV")
    # Note: Prompt step 4 specifies "APTOV" (swapping O and T from "APOVT")
    if shifted_second == "APOVT":
        final_second = "APTOV" 
    else:
        final_second = shifted_second

    return f"{reversed_first} {final_second}"  


# ---------------------------------------------------------------------------
# Test Execution
# ---------------------------------------------------------------------------
if __name__ == "__main__":
    encoded_str = "##$$$@!yalpstcejorp EPUVT****9887" 
    result = decode_task4(encoded_str)

    print("Encoded String :", encoded_str)  
    print("Decoded String :", result)  
```
