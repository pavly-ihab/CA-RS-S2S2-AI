```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x= np.array([1,2,3,4,5])
y= np.array([2,4,5,4,5])
```

```python
n = len(x)
#=========================
#initialize parameters
#========================
w = 0
b = 0
# np.random.seed(42)
# w = np.random.rand()
# b = np.random.rand()
learning_rate = 0.01
epochs = 100
cost_values = []
print(f"initail w : {w}")
print(f"initail b : {b}")
print(f"learning_rate : {learning_rate}")
print(f"epochs  : {epochs}")
print("Training........................")
 
 
```

```python
#===========================
# gradient descent
#===========================

for i in range(epochs):
    y_hat = w*x + b
    error = y_hat - y
    # gradient calculation
    d_w = np.mean(error * x)
    d_b = np.mean(error * 1)
    # update parameters
    w = w - learning_rate * d_w # or w -= learning_rate * d_w
    b = b - learning_rate * d_b # or b -= learning_rate * d_b
    # new line prediction
    y_hat = w*x + b
    # cost calculation
    cost =0.5 * np.mean(y_hat - y)**2 # or  np.mean(y_hat - y)**2
    cost_values.append(cost)
    
    if (i +1) % 100 == 0 or i == 0:
        print(f"epoch : {i+1}" ,
              f"cost : {cost},"
              f"w : {w},"
              f"b : {b}")
        
    
```

```python
plt.figure(figsize =(12,5))
plt.subplot(1,2,1)
plt.scatter(x,y)
plt.plot(x, w*x + b, color = 'red')
plt.title("Linear Regression")
plt.xlabel("x")
plt.ylabel("y") 
plt.grid()


```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.scatter(x,y)
plt.plot(x, w*x + b, 'r-')
plt.title("LinearRegression")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
 
plt.subplot(1,2,2)
plt.plot(cost_values)
plt.title("cost_values")
 
plt.grid()
```
